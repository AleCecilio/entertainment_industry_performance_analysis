### `genres` vs `origina_language`

```python
# Criar a Tabela de Contingência 

top_genres = df_exploded['genres'].value_counts().head(20).index
top_languages = df_exploded['original_language'].value_counts().head(20).index

contingencia_genres_language = pd.crosstab(
    df_exploded['original_language'], 
    df_exploded['genres']
)

display(tb.estilizar_tabela(
    df=contingencia_genres_language.loc[
        top_languages,
        top_genres
    ],
    caption='Tabela de Contingência (Valores Absolutos): original_language vs genres'
))
```

```python
# Executando o Teste do Qui-Quadrado
resultado_genres_language = mt.testar_dependencia_categorica(
    contingencia_genres_language
)

display(tb.estilizar_tabela(
    df=resultado_genres_language,
    caption='Relatório: language_code vs genres'
))
```

# Conclusões da Análise Bivariada:

Esta etapa da Análise Exploratória investigou as relações entre as variáveis estruturais, financeiras e de engajamento do dataset de livros. Diferente de indústrias puramente movidas a marketing, o mercado literário apresentou dinâmicas de engajamento altamente orgânicas.

### Relações Numéricas (N vs N)
* **Engajamento Orgânico:** A matriz de correlação revelou uma relação fortíssima (0.86) entre o total de votos (`total_votes`) e o número de resenhas escritas (`text_reviews_count`). Isso indica que obras populares não apenas acumulam "cliques", mas geram discussão ativa e esforço do leitor.
* **O Mito da "Obra Longa":** Cruzando o número de páginas (`num_pages`) com a nota média (`average_rating`), a correlação é estatisticamente nula (0.03). Um livro maior não garante maior satisfação ou qualidade percebida pelo leitor.

### Categóricas vs Numéricas (C vs N)
* **Arquitetura de Mundo (Gênero vs. Tamanho):** Através de Boxplots, comprovou-se que gêneros como Ficção Científica, Fantasia e Ficção Histórica possuem uma mediana de páginas muito superior a romances e poesias. Esses gêneros exigem uma "arquitetura de mundo" (world-building) que força um formato estruturalmente mais longo.
* **A Bolha do Hype (Popularidade vs. Qualidade):** Ao cruzar a camada de popularidade (`popularity_tier`) com o `global_score`, nota-se que os livros "Mainstream" sofrem maior polarização de notas (violinos mais largos). Obras classificadas como "Cult" ou "Nicho" tendem a ter medianas mais altas, pois agradam de forma cirúrgica o seu público-alvo, escapando do "hate" do público geral.
* **Fronteiras Culturais:** A distribuição de notas por idioma original (`original_language`) indica que obras não-inglesas traduzidas tendem a ter uma régua de aprovação mais alta, sugerindo que apenas a "elite" literária de outros países ganha financiamento para alcançar o mercado global.

###  Relações Categóricas (C vs C)
* **O Foco das Editoras:** Os mapas de calor (Heatmaps) de contingência revelam que as maiores editoras evitam diversificação extrema. A maioria das "Top 10 Publishers" concentra sua produção em nichos específicos (ex: focando 70% de suas obras ativas apenas em Fantasia/Ficção), otimizando suas máquinas de marketing.
